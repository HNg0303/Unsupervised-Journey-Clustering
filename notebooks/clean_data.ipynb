# Clean and Standard Rows and Values in Data.

```python
import pandas as pd
import numpy as np

df = pd.read_csv("data/cleaner_events_log.csv")
```

```python
df.sort_values(by = "timestamp", ascending = True)
```

```python
import json

segmentation_class_name = df["segment_name"].value_counts().index.to_list()
with open("data/segment_name.json", "w", encoding = 'utf-8') as f:
    json.dump(segmentation_class_name, f, indent = 4, ensure_ascii = False)
```

```python
event_name = df["event_type"].value_counts().index.to_list()
with open("data/event_type.json", "w", encoding = 'utf-8') as f:
    json.dump(event_name, f, indent = 4, ensure_ascii = False)
```

```python
screen_name = df["screen_name"].value_counts().index.to_list()
with open("data/screen_name.json", "w", encoding = 'utf-8') as f:
    json.dump(screen_name, f, indent = 4, ensure_ascii = False)
```

```python
activity_duration = df["duration"].value_counts().index.to_list()
with open("data/activity_duration.json", "w", encoding = 'utf-8') as f:
    json.dump(activity_duration, f, indent = 4, ensure_ascii = False)
```

```python
external = df["segmentation.external_id"].value_counts().index.to_list()
with open("data/segmentation_external.json", "w", encoding = 'utf-8') as f:
    json.dump(external, f, indent = 4, ensure_ascii = False)
```

```python
df["event_type"] = df["event_type"].map({
    "[CLY]_view": "View",
    "action": "Action"
})
```

```python
print(len(df["event_type"] == "View"))
df[df["event_type"] == "View"].isna().sum()
```
