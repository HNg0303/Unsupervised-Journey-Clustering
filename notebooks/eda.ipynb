```python
sample_session = (
    df[df["session_id"] == df["session_id"].value_counts().index[0]] #Index is used to label and identify each element of the data -> In this case: the label for each value is the count of each session. Index[0] is the first session_id.
    .sort_values("timestamp")
)

print(f"Len of sample session {session_counts.index[0]} : {len(sample_session)}")

sample_session[
    [
        "session_id",
        "timestamp",
        "key",
        "segmentation.name",
        "segmentation.screen_id",
        "dur",
    ]
].head(50)
```

```python
summary = pd.DataFrame({
    "dtype": df.dtypes.astype(str),
    "non_null": df.notna().sum(),
    "missing": df.isna().sum(),
    "missing_pct": df.isna().mean().mul(100),
    "n_unique": df.nunique(dropna=True)
})

summary["unique_pct"] = (
    summary["n_unique"] / len(df) * 100
)

summary.sort_values("missing_pct", ascending=False)
```
